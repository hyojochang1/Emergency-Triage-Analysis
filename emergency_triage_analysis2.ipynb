```python
bins = [0,19,39,59,79,np.inf]
ages = ['AGE_19','AGE_39','AGE_59','AGE_79','AGE_80+']
Ages = pd.cut(df_clean['Age'],bins=bins, labels= ages, include_lowest=True)

df_clean = df_clean.drop(columns=ages, errors='ignore')
age_dummies = pd.get_dummies(Ages, dtype=int)
loc = df_clean.columns.get_loc('Age') + 1

for i, col in enumerate(age_dummies.columns):
    df_clean.insert(loc + i, col, age_dummies[col])
df_clean.head(4)
```

```python
age_group = pd.cut(df_clean['Age'], bins= bins, include_lowest=True).value_counts().sort_index()
print(age_group)
```

```python
print(df_clean['Sex'].unique())
sex_map = {
    1: 'SEX_Female',
    2: 'SEX_Male'
    
}
sex = df_clean['Sex'].map(sex_map)
sex_dummies = pd.get_dummies(sex, dtype=int)
loc = df_clean.columns.get_loc('Sex') + 1

for i, col in enumerate(sex_dummies.columns):
    df_clean.insert(loc + i, col, sex_dummies[col])

df_clean.head(4)
```

```python
df_clean.tail(4)
```

```python
senior = np.where(df_clean['Age'] >= 65,'Yes','No')
loc = df_clean.columns.get_loc('Age') + 1
df_clean.insert(loc, 'Senior',senior)
df_clean.head(4)

```

```python
(df_clean['Age'] >= 65).map({True: 'Senior', False: 'Non-Senior'}).value_counts().sort_index()
```

```python
ture = df_clean['SpO2'].notna()
low_Spo2 = df_clean['SpO2'].lt(95).where(ture).astype("boolean")
loc = df_clean.columns.get_loc('SpO2') + 1
df_clean.insert(loc, 'Low_SpO2',low_Spo2)
df_clean.head(7)
```

```python
low_Spo2 = df_clean['SpO2'].lt(95).where(ture,'Missing').value_counts()
print(low_Spo2)
```
