```python
elderly = df[df["Age"] >= 65]
print("Number of patients aged 65 of older:",len(elderly))
elderly.head()
```

```python
df["HR"] = pd.to_numeric(df["HR"], errors = "coerce")
high_hr = df[df["HR"] >= 100]
print("Number of patients with HR >= 100:", len(high_hr))
high_hr.head()
```

```python
df.info()
```

```python
df["Saturation"] = pd.to_numeric(df["Saturation"], errors = "coerce")
low_spo2 = df[df["Saturation"] < 95]
print("Number of patients with oxygen saturation below 95:", len(low_spo2))
low_spo2.head(2)
```

```python
df["Saturation"] = pd.to_numeric(df["Saturation"], errors = "coerce")
result = df[(df["Saturation"] < 95) & (df["Age"] >= 65)]
print("Number of senior patients with oxygen saturation below 95:", len(result))
result.head()
```
